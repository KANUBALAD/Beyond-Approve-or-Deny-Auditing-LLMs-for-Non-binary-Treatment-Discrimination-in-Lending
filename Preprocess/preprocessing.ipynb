```python
import pandas as pd
```

### Data preprocess factual and path specific counterfactuals

```python
def preprocess_path_specific_split(data_name, splits):
    if data_name == 'german':
        sens_attributes = ['gender', 'age']
    elif data_name in ('hmda_ny', 'hmda_tx'):
        sens_attributes = ['applicant_sex', 'applicant_race']
    elif data_name == 'homecredit':
        sens_attributes = ['CODE_GENDER', 'age']
    else:
        sens_attributes = ['gender', 'age']
    for split in splits:
        data = pd.read_csv(f"./datasets/{data_name}_{split}.csv")
        factual_cols = [c for c in data.columns if c.endswith('_F') and not c.endswith('_CF')]
        df_factual = data[factual_cols].copy()
        used_attrs = []
        for sens_attribute in sens_attributes:
            if f'{sens_attribute}_CF' in data.columns and f'{sens_attribute}_F' in df_factual.columns:
                idx = df_factual.columns.get_loc(f'{sens_attribute}_F')
                df_factual.drop(columns=[f'{sens_attribute}_F'], inplace=True)
                df_factual.insert(idx, f'{sens_attribute}_cf', data[f'{sens_attribute}_CF'].values)
                used_attrs.append(sens_attribute)
        attr_name = "_".join(used_attrs) if used_attrs else "none"
        data[factual_cols].to_csv(f'./preprocessed_data/{data_name}/{data_name}_factual_{split}.csv', index=False)
        df_factual.to_csv(f'./preprocessed_data/{data_name}/{data_name}_cf_{split}_{attr_name}.csv', index=False)
```

#### German

```python
preprocess_path_specific_split(data_name='german', splits=['train', 'val', 'test'])
```

#### HMDA NY

```python
preprocess_path_specific_split(data_name='hmda_ny', splits=['train', 'val', 'test'])
```

#### HMDA TX

```python
preprocess_path_specific_split(data_name='hmda_tx', splits=['train','val', 'test'])
```

#### HomeCredit

```python
preprocess_path_specific_split(data_name='homecredit', splits=['train', 'val', 'test'])
```
